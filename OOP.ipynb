## 1. Class

### What is a Class?

A class is a blueprint or template used to create objects.

### Real-world analogy

Think of a mobile phone design.

Before manufacturing a mobile phone, a company creates a design that contains:

* Screen size
* Camera
* Battery
* Storage

The design itself is not the actual phone. It is a plan used to create phones.

Similarly, a Python class is a blueprint for creating objects.

## Concept

A class defines:

* what data an object will have
* what actions an object can perform

### Why use it?

Classes group related data (attributes) and behavior (methods) together.

### Syntax

class ClassName:
    # attributes and methods
    pass

```python
class Student:
    name = "Chandini"
    course = "Python"

print(Student.name)
print(Student.course)
```

Key idea: The class describes what a Student object can contain or do.

## 2. Object

An object is an instance of a class.

## Real-world analogy

A mobile phone design can be used to manufacture many actual phones.

Similarly, one class can create many objects.

             Student Class
                   ↓
          ┌────────┼────────┐
          ↓        ↓        ↓
       student1 student2 student3
       Chandini   Rahul     Priya

### Why use objects?

A class is the blueprint; objects are the actual entities created from that blueprint.

```python
class Student:

    def display(self):
        print("This is a Student object.")


student1 = Student()
student2 = Student()

student1.display()
student2.display()
```

## 3. Constructor – __init__()

__init__() is a special method that runs automatically when an object is created.

## Real-world analogy

Imagine you are creating a student ID card.

When a new student joins, you immediately need to enter:

* Name
* Age
* Course

The initial information is set when the student is created.

### Why use it?

It initializes an objects starting data.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Chandini", 23)

print(student.name)
print(student.age)
```

self refers to the current object.

## 4. Instance Variables

Instance variables belong to a particular object. Different objects can have different values.

## Real-world analogy

Consider three students:

* Chandini → Age 23
* Rahul → Age 22
* Priya → Age 21

Each student has their own age.

So age belongs to a particular student.

## Concept

Instance variables are variables that belong to a specific object.

```python
class Student:

    def __init__(self, name, phone):
        self.name = name
        self.phone = phone


student1 = Student("Chandini", 9876543210)
student2 = Student("Reddy", 9876501234)

print(student1.name, student1.phone)
print(student2.name, student2.phone)
```

## 5. Class Variables

A class variable is shared by all objects of a class.

## Real-world analogy

Suppose all students belong to the same college:

* Chandini → ABC College
* Rahul → ABC College
* Priya → ABC College

The college name is common to everyone.

Instead of storing the same information separately for every object, we can create a class variable.

### When to use it?

It can be used when a value should be common to every instance.

```python
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name


student1 = Student("Chandini")
student2 = Student("Rahul")

print(student1.name, student1.college)
print(student2.name, student2.college)
```

## 6. Instance Methods

Instance methods operate on object-specific data and normally receive self as the first parameter.

## Real-world analogy

A student can perform actions:

* attend class
* study
* complete assignment

These actions belong to the student.

### Why?

To perform operations using individual object data.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def study(self):
        print(self.name, "is studying.")


student = Student("Chandini")
student.study()
```

## 7. Static Methods

A static method does not need self or cls. It behaves like a utility function that logically belongs to the class.

Use @staticmethod.

## Real-world analogy

Suppose a college has a utility function:

Check whether a number is positive.

This doesn't depend on any particular student.

It doesn't need student information.

That's a good use case for a static method.

### Why?

To create utility functions related to a class.

### When?

When the method does not need object or class information.

```python
class Calculator:

    @staticmethod
    def multiply(a, b):
        return a * b


print(Calculator.multiply(10, 5))
```

## 8. Class Methods

A class method receives the class itself as cls.

Use @classmethod.

### When to use it?

Common uses include modifying class-level data and creating alternative constructors.

## Real-world analogy

Imagine a college changes its name.

The change applies to the college as a whole, not just one student.

That's class-level behavior.

```python
class Student:

    college = "ABC College"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college


Student.change_college("XYZ College")

print(Student.college)
```
