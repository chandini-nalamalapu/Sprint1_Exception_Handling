## 1. Class

### What is a Class?

A class is a blueprint or template used to create objects.

### Real-world analogy

Think of a mobile phone design.

Before manufacturing a mobile phone, a company creates a design that contains:

* Screen size
* Camera
* Battery
* Storage

The design itself is not the actual phone. It is a plan used to create phones.

Similarly, a Python class is a blueprint for creating objects.

## Concept

A class defines:

* what data an object will have
* what actions an object can perform

### Why use it?

Classes group related data (attributes) and behavior (methods) together.

### Syntax

class ClassName:
    # attributes and methods
    pass

```python
class Student:
    name = "Chandini"
    course = "Python"

print(Student.name)
print(Student.course)
```

Key idea: The class describes what a Student object can contain or do.

## 2. Object

An object is an instance of a class.

## Real-world analogy

A mobile phone design can be used to manufacture many actual phones.

Similarly, one class can create many objects.

             Student Class
                   ↓
          ┌────────┼────────┐
          ↓        ↓        ↓
       student1 student2 student3
       Chandini   Rahul     Priya

### Why use objects?

A class is the blueprint; objects are the actual entities created from that blueprint.

```python
class Student:

    def display(self):
        print("This is a Student object.")


student1 = Student()
student2 = Student()

student1.display()
student2.display()
```

## 3. Constructor – __init__()

__init__() is a special method that runs automatically when an object is created.

## Real-world analogy

Imagine you are creating a student ID card.

When a new student joins, you immediately need to enter:

* Name
* Age
* Course

The initial information is set when the student is created.

### Why use it?

It initializes an objects starting data.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Chandini", 23)

print(student.name)
print(student.age)
```

self refers to the current object.

## 4. Instance Variables

Instance variables belong to a particular object. Different objects can have different values.

## Real-world analogy

Consider three students:

* Chandini → Age 23
* Rahul → Age 22
* Priya → Age 21

Each student has their own age.

So age belongs to a particular student.

## Concept

Instance variables are variables that belong to a specific object.

```python
class Student:

    def __init__(self, name, phone):
        self.name = name
        self.phone = phone


student1 = Student("Chandini", 9876543210)
student2 = Student("Reddy", 9876501234)

print(student1.name, student1.phone)
print(student2.name, student2.phone)
```

## 5. Class Variables

A class variable is shared by all objects of a class.

## Real-world analogy

Suppose all students belong to the same college:

* Chandini → ABC College
* Rahul → ABC College
* Priya → ABC College

The college name is common to everyone.

Instead of storing the same information separately for every object, we can create a class variable.

### When to use it?

It can be used when a value should be common to every instance.

```python
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name


student1 = Student("Chandini")
student2 = Student("Rahul")

print(student1.name, student1.college)
print(student2.name, student2.college)
```

## 6. Instance Methods

Instance methods operate on object-specific data and normally receive self as the first parameter.

## Real-world analogy

A student can perform actions:

* attend class
* study
* complete assignment

These actions belong to the student.

### Why?

To perform operations using individual object data.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def study(self):
        print(self.name, "is studying.")


student = Student("Chandini")
student.study()
```

## 7. Static Methods

A static method does not need self or cls. It behaves like a utility function that logically belongs to the class.

Use @staticmethod.

## Real-world analogy

Suppose a college has a utility function:

Check whether a number is positive.

This doesn't depend on any particular student.

It doesn't need student information.

That's a good use case for a static method.

### Why?

To create utility functions related to a class.

### When?

When the method does not need object or class information.

```python
class Calculator:

    @staticmethod
    def multiply(a, b):
        return a * b


print(Calculator.multiply(10, 5))
```

## 8. Class Methods

A class method receives the class itself as cls.

Use @classmethod.

### When to use it?

Common uses include modifying class-level data and creating alternative constructors.

## Real-world analogy

Imagine a college changes its name.

The change applies to the college as a whole, not just one student.

That's class-level behavior.

```python
class Student:

    college = "ABC College"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college


Student.change_college("XYZ College")

print(Student.college)
```

## 9. Properties

The @property decorator allows us to access a method like an attribute.

Properties are commonly used for controlled access to data.

```python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name


student = Student("Chandini")

print(student.name)
```

## 10. Property Setter

A setter allows us to change the value of a property.

The @name.setter decorator is used to create a setter.

```python
class Student:

    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        self._name = new_name


student = Student("Chandini")

print(student.name)

student.name = "Reddy"

print(student.name)
```

## 11. Encapsulation

Encapsulation means keeping data and methods together inside a class.

It also helps control access to data.

In Python, double underscore is commonly used for private variables.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.__marks)


student = Student("Chandini", 90)

student.display()
```

## 12. Abstraction

Abstraction means hiding unnecessary implementation details.

Python provides abstraction using the abc module.

Abstract classes can contain abstract methods.

```python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog barks")


dog = Dog()

dog.sound()
```

## 13. Inheritance

Inheritance allows a child class to use the properties and methods of a parent class.

The existing class is called the parent class.

The new class is called the child class.

```python
class Parent:

    def show(self):
        print("This is parent class")


class Child(Parent):

    def display(self):
        print("This is child class")


obj = Child()

obj.show()
obj.display()
```

## 14. Multiple Inheritance

Multiple inheritance means one child class inherits from more than one parent class.

```python
class Father:

    def father_property(self):
        print("Father property")


class Mother:

    def mother_property(self):
        print("Mother property")


class Child(Father, Mother):

    def child_property(self):
        print("Child property")


obj = Child()

obj.father_property()
obj.mother_property()
obj.child_property()
```

## 15. Multilevel Inheritance

Multilevel inheritance means inheritance happens through multiple levels.

Example:

Grandparent -> Parent -> Child

```python
class Grandparent:

    def house(self):
        print("Grandparent has a house")


class Parent(Grandparent):

    def car(self):
        print("Parent has a car")


class Child(Parent):

    def bike(self):
        print("Child has a bike")


obj = Child()

obj.house()
obj.car()
obj.bike()
```

## 16. Hierarchical Inheritance

Hierarchical inheritance means multiple child classes inherit from the same parent class.

```python
class Parent:

    def show(self):
        print("This is parent class")


class Child1(Parent):

    def display1(self):
        print("This is child 1")


class Child2(Parent):

    def display2(self):
        print("This is child 2")


obj1 = Child1()

obj1.show()
obj1.display1()


obj2 = Child2()

obj2.show()
obj2.display2()
```

## 17. Polymorphism

Polymorphism means one method name can have different behaviors.

Different classes can have the same method name with different implementations.

```python
class Dog:

    def sound(self):
        print("Dog barks")


class Cat:

    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

## 18. Method Overriding

Method overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.

```python
class Parent:

    def show(self):
        print("Parent method")


class Child(Parent):

    def show(self):
        print("Child method")


obj = Child()

obj.show()
```

## 19. Method Overloading – Python Approach 

Python does not support traditional method overloading like Java.

We can achieve similar behavior using default arguments or *args.

Here, the same method can work with different numbers of arguments.

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c


calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
```

# 20. Composition

Composition represents a strong "has-a" relationship.

One class contains an object of another class.

The contained object is usually created inside the main class.

```python
class Engine:

    def start(self):
        print("Engine started")


class Car:

    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")


car = Car()

car.start_car()
```

# 21. Aggregation

Aggregation represents a weak "has-a" relationship.

One class uses an object of another class, but the object can exist independently.

```python
class Student:

    def __init__(self, name):
        self.name = name


class College:

    def __init__(self, student):
        self.student = student


student = Student("Chandini")

college = College(student)

print(college.student.name)
```

# 22. Association

Association means two classes are connected with each other, but they can exist independently.

Example: A Teacher can teach a Student, but both Teacher and Student can exist separately.

```python
class Student:
    def __init__(self, name):
        self.name = name


class Teacher:
    def __init__(self, name):
        self.name = name

    def teach(self, student):
        print(self.name, "is teaching", student.name)


student1 = Student("Chandini")
teacher1 = Teacher("Ravi")

teacher1.teach(student1)
```

# 23. Magic (Dunder) Methods

Magic methods are special methods in Python.

They start and end with double underscores.

Examples:
__init__()
__str__()
__len__()

```python
class Student:
    def __init__(self, name):
        self.name = name


student1 = Student("Chandini")

print(student1.name)
```

# 24. super()

super() is used to call the parent class constructor or method.

It is commonly used in inheritance.

```python
class Parent:
    def __init__(self):
        print("This is Parent class")


class Child(Parent):
    def __init__(self):
        super().__init__()
        print("This is Child class")


child1 = Child()
```

# 25. __str__()

__str__() is used to return a readable string representation of an object.

It is automatically called when we use print() with an object.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return "Student name is " + self.name + " and age is " + str(self.age)


student1 = Student("Chandini", 23)

print(student1)
```

# 26. __repr__()

__repr__() gives a developer-friendly representation of an object.

It is useful for understanding the object and its values.    

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


student1 = Student("Chandini", 23)

print(repr(student1))
```

# 27. __len__()

__len__() is used to define the length of an object.

It is automatically called when we use len() with an object.

```python
class Team:
    def __init__(self, players):
        self.players = players

    def __len__(self):
        return len(self.players)


team1 = Team(["Ravi", "Chandini", "Rahul", "Priya"])

print(len(team1))
```

# 28. Dataclasses – Bonus

Dataclasses are used to create classes easily.

They automatically provide methods like __init__() and __repr__().

Dataclasses are useful for storing data.

```python
from dataclasses import dataclass


@dataclass
class Student:
    name: str
    age: int
    city: str


student1 = Student("Chandini", 23, "Hyderabad")

print(student1)
```

# 29. Dataclass Example

We can use dataclasses to store employee information.

```python
from dataclasses import dataclass


@dataclass
class Employee:
    name: str
    salary: int


employee1 = Employee("Chandini", 10000)

print(employee1.name)
print(employee1.salary)
```

# 30. Singleton Pattern – Bonus

Singleton Pattern means only one object of a class is created.

If we create the object again, it returns the same object.

```python
class Database:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            print("Database object created")

        return cls._instance


db1 = Database()
db2 = Database()

print(db1 is db2)
```
