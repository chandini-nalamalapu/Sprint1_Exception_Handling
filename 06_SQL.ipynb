## Database Design

In this module, I will learn SQL and basic database operations.

### Database

A database is a collection of related data that is stored in an organized way.

Example:

A College Database can store information about students, teachers, courses, and marks.

```python
import mysql.connector
```

```python
password = input("Enter your MySQL password: ")

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password=password
)

cursor = connection.cursor()

print("MySQL connected successfully")
```

### Create Database

We will create a new database called CollegeDB.

```python
cursor.execute("DROP DATABASE IF EXISTS CollegeDB")
cursor.execute("CREATE DATABASE CollegeDB")
cursor.execute("USE CollegeDB")

print("CollegeDB created and selected")
```

### Tables

A table stores data in rows and columns.

Example:

Students table stores information about students.

```python
query = """
CREATE TABLE Students (
    Student_ID INT PRIMARY KEY,
    Name VARCHAR(50),
    Age INT,
    City VARCHAR(50),
    Marks INT
)
"""

cursor.execute(query)

print("Students table created")
```

```python
query = """
INSERT INTO Students (Student_ID, Name, Age, City, Marks)
VALUES
(1, 'Anjali', 20, 'Hyderabad', 85),
(2, 'Rahul', 21, 'Chennai', 90),
(3, 'Suresh', 19, 'Hyderabad', 75),
(4, 'Divya', 22, 'Mumbai', 88),
(5, 'Karthik', 20, 'Pune', 92)
"""

cursor.execute(query)
connection.commit()

print("Student data inserted")
```

### Primary Key

A primary key uniquely identifies each row in a table.

It cannot contain duplicate or NULL values.

Here, Student_ID is the primary key.

### Foreign Key

A foreign key connects one table with another table.

It refers to the primary key of another table.

```python
query = """
CREATE TABLE Marks (
    Mark_ID INT PRIMARY KEY,
    Student_ID INT,
    Subject VARCHAR(50),
    Marks INT,
    FOREIGN KEY (Student_ID) REFERENCES Students(Student_ID)
)
"""

cursor.execute(query)

print("Marks table created")
```

```python
query = """
INSERT INTO Marks (Mark_ID, Student_ID, Subject, Marks)
VALUES
(101, 1, 'Python', 85),
(102, 2, 'Python', 90),
(103, 3, 'Python', 75),
(104, 4, 'Python', 88),
(105, 5, 'Python', 92)
"""

cursor.execute(query)
connection.commit()

print("Marks data inserted")
```

### Relationships

A relationship is a connection between two or more tables using keys.

Example:

One student can have marks.

Students and Marks tables are connected using Student_ID.

### Constraints

Constraints are rules applied to table columns to make sure the data is accurate and valid.

Common constraints:

- NOT NULL
- UNIQUE
- PRIMARY KEY
- FOREIGN KEY
- DEFAULT
- CHECK

```python
query = """
CREATE TABLE Courses (
    Course_ID INT PRIMARY KEY,
    Course_Name VARCHAR(50) NOT NULL,
    Course_Code VARCHAR(20) UNIQUE,
    Duration INT DEFAULT 6,
    CHECK (Duration > 0)
)
"""

cursor.execute(query)

print("Courses table created")
```

### SELECT

SELECT is used to retrieve data from a table.

```python
cursor.execute("SELECT Name, Age FROM Students")

for row in cursor.fetchall():
    print(row)
```

### WHERE

WHERE is used to filter records based on a condition.

```python
cursor.execute("SELECT * FROM Students WHERE Marks > 80")

for row in cursor.fetchall():
    print(row)
```

### ORDER BY

ORDER BY is used to sort the result in ascending or descending order.

```python
cursor.execute("SELECT * FROM Students ORDER BY Age ASC")

for row in cursor.fetchall():
    print(row)
```

### GROUP BY

GROUP BY is used to group rows that have the same values in a column.

```python
cursor.execute("""
SELECT City, COUNT(*)
FROM Students
GROUP BY City
""")

for row in cursor.fetchall():
    print(row)
```

### HAVING

HAVING is used to filter grouped data after using GROUP BY.

```python
cursor.execute("""
SELECT City, COUNT(*)
FROM Students
GROUP BY City
HAVING COUNT(*) >= 1
""")

for row in cursor.fetchall():
    print(row)
```

### DISTINCT

DISTINCT is used to return only unique values by removing duplicates.

```python
cursor.execute("SELECT DISTINCT City FROM Students")

for row in cursor.fetchall():
    print(row)
```

### LIMIT

LIMIT is used to restrict the number of rows returned.

```python
cursor.execute("SELECT * FROM Students LIMIT 3")

for row in cursor.fetchall():
    print(row)
```

### Joins

Joins are used to combine data from two or more tables.

### INNER JOIN

INNER JOIN returns only the rows that have matching values in both tables.

```python
cursor.execute("""
SELECT Students.Name, Marks.Subject, Marks.Marks
FROM Students
INNER JOIN Marks
ON Students.Student_ID = Marks.Student_ID
""")

for row in cursor.fetchall():
    print(row)
```

### LEFT JOIN

LEFT JOIN returns all rows from the left table and matching rows from the right table.

```python
cursor.execute("""
SELECT Students.Name, Marks.Marks
FROM Students
LEFT JOIN Marks
ON Students.Student_ID = Marks.Student_ID
""")

for row in cursor.fetchall():
    print(row)
```

### RIGHT JOIN

RIGHT JOIN returns all rows from the right table and matching rows from the left table.

```python
cursor.execute("""
SELECT Students.Name, Marks.Marks
FROM Students
RIGHT JOIN Marks
ON Students.Student_ID = Marks.Student_ID
""")

for row in cursor.fetchall():
    print(row)
```

### FULL JOIN

FULL JOIN returns all matching and non-matching rows from both tables.

MySQL does not directly support FULL JOIN, so we can use LEFT JOIN and RIGHT JOIN with UNION.

```python
cursor.execute("""
SELECT Students.Name, Marks.Marks
FROM Students
LEFT JOIN Marks
ON Students.Student_ID = Marks.Student_ID

UNION

SELECT Students.Name, Marks.Marks
FROM Students
RIGHT JOIN Marks
ON Students.Student_ID = Marks.Student_ID
""")

for row in cursor.fetchall():
    print(row)
```

### CROSS JOIN

CROSS JOIN returns every possible combination of rows from both tables.

```python
cursor.execute("""
SELECT Students.Name, Marks.Subject
FROM Students
CROSS JOIN Marks
""")

for row in cursor.fetchall():
    print(row)
```

### Advanced SQL
### UNION

UNION combines the results of two or more SELECT queries and removes duplicate rows.

```python
cursor.execute("""
SELECT Name FROM Students
UNION
SELECT Subject FROM Marks
""")

for row in cursor.fetchall():
    print(row)
```

### CASE

CASE is used to apply conditions and return different values based on those conditions.

```python
cursor.execute("""
SELECT Name, Marks,
CASE
    WHEN Marks >= 90 THEN 'Excellent'
    WHEN Marks >= 80 THEN 'Good'
    ELSE 'Average'
END AS Result
FROM Students
""")

for row in cursor.fetchall():
    print(row)
```

### Subqueries

A subquery is a query written inside another SQL query.

```python
cursor.execute("""
SELECT Name, Marks
FROM Students
WHERE Marks > (
    SELECT AVG(Marks)
    FROM Students
)
""")

for row in cursor.fetchall():
    print(row)
```

### Common Table Expressions (CTE)

A CTE is a temporary result that can be used by another query.

```python
cursor.execute("""
WITH HighMarks AS (
    SELECT Name, Marks
    FROM Students
    WHERE Marks > 80
)
SELECT * FROM HighMarks
""")

for row in cursor.fetchall():
    print(row)
```

### Views

A view is a virtual table created from the result of a SQL query.

```python
cursor.execute("""
CREATE OR REPLACE VIEW StudentDetails AS
SELECT Name, City, Marks
FROM Students
""")

print("View created")
```

```python
cursor.execute("SELECT * FROM StudentDetails")

for row in cursor.fetchall():
    print(row)
```

### Indexes

An index is used to make searching data faster.

```python
cursor.execute("""
CREATE INDEX idx_student_city
ON Students(City)
""")

print("Index created")
```

### Window Functions

Window functions perform calculations across a set of rows without grouping them.

### ROW_NUMBER()

ROW_NUMBER() gives a unique number to each row.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    ROW_NUMBER() OVER (ORDER BY Marks DESC) AS student_number
FROM Students
""")

for row in cursor.fetchall():
    print(row)
```

### RANK()

RANK() gives the same rank to rows with the same value.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    RANK() OVER (ORDER BY Marks DESC) AS Student_Rank
FROM Students
""")

for row in cursor.fetchall():
    print(row)
```

### DENSE_RANK()

DENSE_RANK() gives ranks without skipping numbers after a tie.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    DENSE_RANK() OVER (ORDER BY Marks DESC) AS Student_Rank
FROM Students
""")

for row in cursor.fetchall():
    print(row)
```

### LAG()

LAG() is used to access the value from the previous row.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    LAG(Marks) OVER (ORDER BY Student_ID) AS Previous_Marks
FROM Students
""")

for row in cursor.fetchall():
    print(row)
```

### LEAD()

LEAD() is used to access the value from the next row.

```python
cursor.execute("""
SELECT
    Name,
    Marks,
    LEAD(Marks) OVER (ORDER BY Student_ID) AS Next_Marks
FROM Students
""")

for row in cursor.fetchall():
    print(row)
```
