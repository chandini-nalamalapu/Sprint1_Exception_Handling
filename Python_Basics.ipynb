# Basic Concepts:

### 1. Variable

A variable is a named container used to store or refer to a value in a program. It allows us to save information and use that information later.

In Python, we do not need to declare the variable type separately. Python identifies the data type from the value assigned to the variable.

### Real-world analogy

Think of a variable like a labelled box.

- A box labelled `name` can contain `"Chandini"`.
- A box labelled `age` can contain `23`.
- If the value changes, we can put a new value into the same labelled box.

Variables are useful because values can be reused and changed while a program is running.

```python
name = "Chandini"
age = 23
city = "Hyderabad"

print(name)
print(age)
print(city)
```

### 2. Data Types

A Data Type tells Python what kind of data is contained in a variable.

In Python, the data type is set when we assign a value to a variable.

Common Python data types are:

| Data Type | Example | Purpose |
|---|---|---|
| `int` | `25` | Whole numbers |
| `float` | `25.5` | Decimal numbers |
| `str` | `"Python"` | Text |
| `bool` | `True` | True/False values |
| `list` | `[10, 20, 30]` | Ordered, changeable collection |
| `tuple` | `(10, 20, 30)` | Ordered collection |
| `set` | `{10, 20, 30}` | Collection of unique values |
| `dict` | `{"name": "Alex"}` | Key-value pairs |

### type()

The `type()` function is used to check the data type of a variable.

```python
student_name = "Chandini"
student_age = 23
student_marks = 85.5
is_passed = True

print(type(student_name))
print(type(student_age))
print(type(student_marks))
print(type(is_passed))
```

### 3. Type Casting

Type casting means converting a value from one data type to another.

Python provides functions such as:

- `int()` - converts a value into an integer
- `float()` - converts a value into a decimal number
- `str()` - converts a value into a string
- `bool()` - converts a value into a Boolean value

```python
number = 50

number_string = str(number)

print(number)
print(number_string)

print(type(number))
print(type(number_string))
```

    The value looks the same, but the data types are different.

    number is an integer and number_string is a string.

### 4. Operators

Operators are symbols or keywords used to perform operations on values.

Python has different types of operators:

1. Arithmetic operators
2. Comparison operators
3. Logical operators
4. Assignment operators
5. Membership operators
6. Identity operators

#### 4.1 Arithmetic Operators

Arithmetic operators are used for calculations.

The main arithmetic operators are:

`+`, `-`, `*`, `/`, `//`, `%`, `**`

```python
a = 20
b = 6

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
```

#### 4.2 Comparison Operators

Comparison operators are used to compare values.

The main comparison operators are:

`==`, `!=`, `>`, `<`, `>=`, `<=`

They return a Boolean result such as `True` or `False`.

```python
a = 15
b = 10

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

#### 4.3 Logical Operators

Logical operators are used to combine conditions.

The main logical operators are:

- `and`
- `or`
- `not`

```python
age = 22
marks = 75

print(age > 18 and marks > 50)
print(age < 18 or marks > 50)
print(not(age > 18))
```

#### 4.4 Assignment Operators

Assignment operators are used to assign or update values.

Common assignment operators are:

`=`, `+=`, `-=`, `*=`, `/=`

```python
score = 20

score += 5
print(score)

score -= 3
print(score)

score *= 2
print(score)
```

#### 4.5 Membership Operators

Membership operators are used to check whether a value is present in an object.

The membership operators are:

- `in`
- `not in`

```python
languages = ["Python", "Java", "C++", "JavaScript"]

print("Python" in languages)
print("PHP" not in languages)
```

#### 4.6 Identity Operators

Identity operators are used to check whether two variables refer to the same object or memory location.

The identity operators are:

- `is`
- `is not`

```python
numbers = [10, 20, 30]

numbers1 = numbers
numbers2 = [10, 20, 30]

print(numbers is numbers1)
print(numbers is numbers2)
```

### 5. Input and Output

Input means taking information from the user into the program.

Output means displaying information from the program to the user.

In Python:

- `input()` is used to take input from the user.
- `print()` is used to display output.

```python
name = input("Enter your name: ")

print("Welcome", name)
```

### 6. Comments

Comments are notes written inside source code for humans to understand the program.

Python does not execute normal comments as program instructions.

Comments start with `#`, and Python will ignore them.

```python
# This is a comment

name = "Chandini"

print("Hello", name)
```

### Multi-line Comments

Python does not have a separate multi-line comment syntax.

We can use `#` for each line.

```python
# This is the first comment
# This is the second comment
# This is the third comment

print("Learning Python")
```

Alternatively, we can use a multiline string.

Since Python ignores unassigned string literals, multiline strings can be used to add longer comments to the code.

"""
This is a multiline comment.
It contains more than one line.
It is used to explain the code.
"""

print("Python Fundamentals")

## Control Flow

Control flow is used to control the order in which statements are executed in a Python program.

### 7. IF / ELSE / ELIF

`if`, `elif`, and `else` help Python make decisions.

- `if` – Executes a block of code when a condition is true.
- `elif` – Checks another condition if the previous condition is false.
- `else` – Executes a block of code when none of the conditions are true.

### Real-world analogy

- If it is raining, take an umbrella.
- Else if it is sunny, wear sunglasses.
- Else, carry on normally.

Python checks each condition one by one. As soon as one condition is true, it performs that action.

```python
marks = int(input("Enter your marks: "))

if marks < 35:
    print("Fail")
elif marks == 35:
    print("Just Passed")
else:
    print("Passed")
```
