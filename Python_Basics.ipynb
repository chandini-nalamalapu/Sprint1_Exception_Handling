# Basic Concepts:

### 1. Variable

A variable is a named container used to store or refer to a value in a program. It allows us to save information and use that information later.

In Python, we do not need to declare the variable type separately. Python identifies the data type from the value assigned to the variable.

### Real-world analogy

Think of a variable like a labelled box.

- A box labelled `name` can contain `"Chandini"`.
- A box labelled `age` can contain `23`.
- If the value changes, we can put a new value into the same labelled box.

Variables are useful because values can be reused and changed while a program is running.

```python
name = "Chandini"
age = 23
city = "Hyderabad"

print(name)
print(age)
print(city)
```

### 2. Data Types

A Data Type tells Python what kind of data is contained in a variable.

In Python, the data type is set when we assign a value to a variable.

Common Python data types are:

| Data Type | Example | Purpose |
|---|---|---|
| `int` | `25` | Whole numbers |
| `float` | `25.5` | Decimal numbers |
| `str` | `"Python"` | Text |
| `bool` | `True` | True/False values |
| `list` | `[10, 20, 30]` | Ordered, changeable collection |
| `tuple` | `(10, 20, 30)` | Ordered collection |
| `set` | `{10, 20, 30}` | Collection of unique values |
| `dict` | `{"name": "Alex"}` | Key-value pairs |

### type()

The `type()` function is used to check the data type of a variable.

```python
student_name = "Chandini"
student_age = 23
student_marks = 85.5
is_passed = True

print(type(student_name))
print(type(student_age))
print(type(student_marks))
print(type(is_passed))
```

### 3. Type Casting

Type casting means converting a value from one data type to another.

Python provides functions such as:

- `int()` - converts a value into an integer
- `float()` - converts a value into a decimal number
- `str()` - converts a value into a string
- `bool()` - converts a value into a Boolean value

```python
number = 50

number_string = str(number)

print(number)
print(number_string)

print(type(number))
print(type(number_string))
```

    The value looks the same, but the data types are different.

    number is an integer and number_string is a string.

### 4. Operators

Operators are symbols or keywords used to perform operations on values.

Python has different types of operators:

1. Arithmetic operators
2. Comparison operators
3. Logical operators
4. Assignment operators
5. Membership operators
6. Identity operators

#### 4.1 Arithmetic Operators

Arithmetic operators are used for calculations.

The main arithmetic operators are:

`+`, `-`, `*`, `/`, `//`, `%`, `**`

```python
a = 20
b = 6

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
```

#### 4.2 Comparison Operators

Comparison operators are used to compare values.

The main comparison operators are:

`==`, `!=`, `>`, `<`, `>=`, `<=`

They return a Boolean result such as `True` or `False`.

```python
a = 15
b = 10

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

#### 4.3 Logical Operators

Logical operators are used to combine conditions.

The main logical operators are:

- `and`
- `or`
- `not`

```python
age = 22
marks = 75

print(age > 18 and marks > 50)
print(age < 18 or marks > 50)
print(not(age > 18))
```

#### 4.4 Assignment Operators

Assignment operators are used to assign or update values.

Common assignment operators are:

`=`, `+=`, `-=`, `*=`, `/=`

```python
score = 20

score += 5
print(score)

score -= 3
print(score)

score *= 2
print(score)
```

#### 4.5 Membership Operators

Membership operators are used to check whether a value is present in an object.

The membership operators are:

- `in`
- `not in`

```python
languages = ["Python", "Java", "C++", "JavaScript"]

print("Python" in languages)
print("PHP" not in languages)
```

#### 4.6 Identity Operators

Identity operators are used to check whether two variables refer to the same object or memory location.

The identity operators are:

- `is`
- `is not`

```python
numbers = [10, 20, 30]

numbers1 = numbers
numbers2 = [10, 20, 30]

print(numbers is numbers1)
print(numbers is numbers2)
```

### 5. Input and Output

Input means taking information from the user into the program.

Output means displaying information from the program to the user.

In Python:

- `input()` is used to take input from the user.
- `print()` is used to display output.

```python
name = input("Enter your name: ")

print("Welcome", name)
```

### 6. Comments

Comments are notes written inside source code for humans to understand the program.

Python does not execute normal comments as program instructions.

Comments start with `#`, and Python will ignore them.

```python
# This is a comment

name = "Chandini"

print("Hello", name)
```

### Multi-line Comments

Python does not have a separate multi-line comment syntax.

We can use `#` for each line.

```python
# This is the first comment
# This is the second comment
# This is the third comment

print("Learning Python")
```

Alternatively, we can use a multiline string.

Since Python ignores unassigned string literals, multiline strings can be used to add longer comments to the code.

"""
This is a multiline comment.
It contains more than one line.
It is used to explain the code.
"""

print("Python Fundamentals")

## Control Flow

Control flow is used to control the order in which statements are executed in a Python program.

### 7. IF / ELSE / ELIF

`if`, `elif`, and `else` help Python make decisions.

- `if` – Executes a block of code when a condition is true.
- `elif` – Checks another condition if the previous condition is false.
- `else` – Executes a block of code when none of the conditions are true.

### Real-world analogy

- If it is raining, take an umbrella.
- Else if it is sunny, wear sunglasses.
- Else, carry on normally.

Python checks each condition one by one. As soon as one condition is true, it performs that action.

```python
marks = int(input("Enter your marks: "))

if marks < 35:
    print("Fail")
elif marks == 35:
    print("Just Passed")
else:
    print("Passed")
```

# 8. for loop

A for loop is used to repeat a block of code for each item in a sequence.

We can use a for loop with lists, strings, tuples and ranges.

Example:
Print numbers from 1 to 5.

```python
for number in range(1, 6):
    print(number)
```

# 9. while loop

A while loop is used to repeat a block of code as long as a condition is true.

The loop stops when the condition becomes false.

```python
number = 1

while number <= 5:
    print(number)
    number = number + 1
```

# 10. break

The break statement is used to stop a loop immediately.

When Python reaches break, the loop will end.

```python
for number in range(1, 10):
    if number == 5:
        break
    print(number)
```

# 11. continue

The continue statement is used to skip the current iteration of a loop.

The loop continues with the next iteration.

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

# 12. pass

The pass statement does nothing.

It is used when we need to write a block of code but do not want to add any code yet.

```python
for number in range(1, 6):
    if number == 3:
        pass
    print(number)
```

# Collections

# 13. List

A list is used to store multiple values in a single variable.

Lists are ordered and changeable.

Lists can contain duplicate values.

Lists are written using square brackets [].

```python
fruits = ["Apple", "Banana", "Mango"]

print(fruits)
print(fruits[0])

fruits.append("Orange")

print(fruits)
```

# 14. Tuple

A tuple is used to store multiple values in a single variable.

Tuples are ordered but cannot be changed after creation.

Tuples are written using round brackets ().

```python
colors = ("Red", "Green", "Blue")

print(colors)
print(colors[0])
print(colors[2])
```

# 15. Set

A set is used to store multiple unique values.

Sets do not allow duplicate values.

Sets are written using curly brackets {}.

```python
numbers = {10, 20, 30, 20, 10}

print(numbers)

numbers.add(40)

print(numbers)
```

# 16. Dictionary

A dictionary stores data in key-value pairs.

Each key is used to access its value.

Dictionaries are written using curly brackets {}.

```python
student = {
    "name": "Chandini",
    "age": 23,
    "course": "Python"
}

print(student)
print(student["name"])
print(student["course"])
```

# 17. List Comprehension

List comprehension is a short way to create a new list.

It can make list creation easier and shorter.

```python
numbers = [1, 2, 3, 4, 5]

squares = [number * number for number in numbers]

print(squares)
```

# 18. Dictionary Comprehension

Dictionary comprehension is a short way to create a dictionary.

It creates key-value pairs in a simple way.

```python
numbers = [1, 2, 3, 4, 5]

squares = {number: number * number for number in numbers}

print(squares)
```

## Functions


# 19. Function Definition

A function is a block of reusable code.

We define a function using the def keyword.

A function runs when we call it.

```python
def greet():
    print("Hello Chandini!, welcome to Python!")

greet()
```

# 20. Arguments

Arguments are values passed to a function when the function is called.

They allow us to send data into a function.

```python
def greet(name):
    print("Hello", name)

greet("Chandini")
```

# 21. Default Arguments

A default argument has a default value.

If we do not provide a value while calling the function, Python uses the default value.

```python
def greet(name="Chandini"):
    print("Hello", name)

greet()
greet("Rahul")
```

# 22. Keyword Arguments

Keyword arguments are passed to a function using the parameter name.

This allows us to provide arguments by name.

```python
def student_info(name, age):
    print("Name:", name)
    print("Age:", age)

student_info(age=23, name="Chandini")
```

# 23. Variable-Length Arguments

Variable-length arguments allow a function to accept multiple values.

*args is used for multiple positional arguments.

**kwargs is used for multiple keyword arguments.

```python
def add_numbers(*args):
    total = 0

    for number in args:
        total = total + number

    print("Total:", total)

add_numbers(10, 20, 30, 40)
```

```python
def student_details(**kwargs):
    print(kwargs)

student_details(name="Chandini", age=23, course="Python")
```

# 24. Lambda Functions

A lambda function is a small anonymous function.

It is written using the lambda keyword.

Lambda functions are usually used for simple operations.

```python
square = lambda number: number * number

print(square(5))
```

# 25. Recursive Functions

A recursive function is a function that calls itself.

A stopping condition is required to prevent the function from running forever.

Example:
Calculate the factorial of a number.

```python
def factorial(number):
    if number == 1:
        return 1
    else:
        return number * factorial(number - 1)

print(factorial(5))
```

 # Advanced Python


# 26. Iterators

An iterator is an object that allows us to access elements one at a time.

The iter() function creates an iterator.

The next() function gets the next value.

```python
numbers = [10, 20, 40]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

# 27. Generators

A generator is a special type of function that produces values one at a time.

Generators use the yield keyword.

They are useful when we do not want to store all values in memory at once.

```python
def numbers():
    yield 1
    yield 2
    yield 3

for number in numbers():
    print(number)
```

# 28. yield

The yield keyword is used inside a generator function.

It returns a value one at a time and pauses the function.

When the next value is requested, the function continues from where it stopped.

```python
def count_numbers():
    yield 1
    yield 2
    yield 3

result = count_numbers()

print(next(result))
print(next(result))
print(next(result))
```

# 29. Decorators

A decorator is used to add extra functionality to an existing function without changing the original function.

Decorators use the @ symbol.

```python
def my_decorator(function):
    def wrapper():
        print("Before function")
        function()
        print("After function")

    return wrapper


@my_decorator
def greet():
    print("Hello")


greet()
```

# 30. Closures

A closure is a function that remembers values from its outer function even after the outer function has finished.

```python
def outer_function(message):

    def inner_function():
        print(message)

    return inner_function


greet = outer_function("Hello Chandini")

greet()
```

# 31. global

The global keyword is used when we want to change a global variable inside a function.

```python
count = 10

def change_count():
    global count
    count = 20

change_count()

print(count)
```

# 32. nonlocal

The nonlocal keyword is used inside a nested function to change a variable from the outer function.

```python
def outer():
    number = 10

    def inner():
        nonlocal number
        number = 20

    inner()

    print(number)


outer()
```

# Built-in Functions


# 37. map()

The map() function applies a function to every item in an iterable.

It is commonly used to perform the same operation on multiple values.

```python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda number: number * number, numbers))

print(squares)
```

# 38. filter()

The filter() function is used to select items from an iterable based on a condition.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda number: number % 2 == 0, numbers))

print(even_numbers)
```

# 39. zip()

The zip() function combines elements from two or more iterables.

It pairs the elements based on their positions.

```python
names = ["Chandini", "Rahul", "Priya"]
ages = [23, 24, 22]

students = list(zip(names, ages))

print(students)
```

# 40. enumerate()

The enumerate() function gives both the index and the value while looping through a sequence.

```python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

# 41. sorted()

The sorted() function returns a sorted version of an iterable.

By default, it sorts values in ascending order.

```python
numbers = [50, 10, 40, 20, 30]

sorted_numbers = sorted(numbers)

print(sorted_numbers)
```

# 42. any()

The any() function returns True if at least one value in an iterable is True.

It returns False if all values are False.

```python
numbers = [1, 3, 5, 8]

result = any(number % 2 == 0 for number in numbers)

print(result)
```

# 43. all()

The all() function returns True if all values in an iterable are True.

It returns False if at least one value is False.

```python
numbers = [2, 4, 6, 8]

result = all(number % 2 == 0 for number in numbers)

print(result)
```

# 44. isinstance()

The isinstance() function checks whether a value belongs to a particular data type.

It returns True or False.

```python
name = "Chandini"
age = 23

print(isinstance(name, str))
print(isinstance(age, int))
```

# 45. id()

The id() function returns the unique identity of an object during its lifetime.

It is mainly useful for understanding how Python handles objects in memory.

```python
name = "Chandini"

print(id(name))
```
