## 1) READING TEXT FILES

Reading a text file means getting the data stored in a file.

Python uses the open() function with read mode "r".

Methods:

read() - Reads the complete file

readline() - Reads one line

readlines() - Reads all lines as a list

```python
# Creating a text file for reading

file = open("notes.txt", "w")

file.write("Python File Handling\n")
file.write("Reading files is easy\n")
file.write("This is my practice file")

file.close()

print("File created")
```

```python
# Reading one line from the file

file = open("notes.txt", "r")

line = file.readline()

print(line)

file.close()
```

```python
# Reading all lines as a list

file = open("notes.txt", "r")

lines = file.readlines()

print(lines)

file.close()
```

## 2) WRITING TEXT FILES

Writing means storing information inside a text file.

Modes:

w - Write mode. It creates a new file or overwrites existing data.

a - Append mode. It adds data at the end of the file.

x - Create mode. It creates a new file.

```python
# Writing data into a file

file = open("student_info.txt", "w")

file.write("Name: Chandini\n")
file.write("Course: Python\n")
file.write("Module: File Handling")

file.close()

print("Student information saved")
```

```python
# Adding new information to the existing file

file = open("student_info.txt", "a")

file.write("\nStatus: Learning")

file.close()

print("New information added")
```

```python
# Creating a new file

file = open("new_file.txt", "x")

file.write("This is a newly created file.")

file.close()

print("New file created")
```

## 4) JSON FILES

JSON stands for JavaScript Object Notation.

JSON stores information in key-value pairs.

JSON files are commonly used with APIs and configuration files.

```python
import json

student = {
    "name": "Rahul",
    "age": 23,
    "course": "Python",
    "city": "Guntur"
}
with open("student_data.json", "w") as file:
    json.dump(student, file)

print("JSON file created")
```

```python
import json

with open("student_data.json", "r") as file:
    student_data = json.load(file)

print(student_data)
```

```python
print("Name:", student_data["name"])
print("Course:", student_data["course"])
print("City:", student_data["city"])
```

## 5) PICKLE FILES

Pickle is used to store Python objects in binary format.

Serialization means saving a Python object.

Deserialization means loading a saved Python object.

```python
import pickle

employee = {
    "name": "Kiran",
    "age": 25,
    "department": "IT",
    "salary": 30000
}

with open("employee.pkl", "wb") as file:
    pickle.dump(employee, file)

print("Employee data saved")
```

```python
import pickle

with open("employee.pkl", "rb") as file:
    employee_data = pickle.load(file)

print(employee_data)
```

## 6) CONTEXT MANAGER (with)

The with statement is used to work with files safely.

It automatically opens the file and closes the file after use.

It is easier than opening and closing the file manually.

```python
    # Writing a file using with

with open("message.txt", "w") as file:
    file.write("Good morning\n")
    file.write("Welcome to Python File Handling")

print("Message saved")
```

```python
# Reading a file using with

with open("message.txt", "r") as file:
    message = file.read()

print(message)
```

## 7) PATHLIB

pathlib provides an object-oriented way to work with file and folder paths.

It is easy to use and works across different operating systems.

```python
from pathlib import Path

file_path = Path("report.txt")

if file_path.exists():
    print("The file exists")
else:
    print("The file does not exist")
```

```python
from pathlib import Path

file_path = Path("report.txt")

file_path.write_text("This file was created using pathlib.")

print(file_path.read_text())
```

## 9) SHUTIL MODULE

shutil stands for Shell Utilities.

It provides functions for high-level file and folder operations.

Common functions:

copy() - Copies a file
move() - Moves a file
rmtree() - Removes a folder and its contents
copytree() - Copies a complete folder

```python
import shutil

# Create a file first
with open("original.txt", "w") as file:
    file.write("This is the original file.")

# Copy the file
shutil.copy("original.txt", "backup.txt")

print("File copied successfully")
```

```python
import shutil

shutil.move("backup.txt", "moved_backup.txt")

print("File moved successfully")
```

## 10) FILE VALIDATION

File validation checks a file before using it.

We can check:

Whether the file exists

Whether the file can be read

Whether the file can be written

Whether the file is empty
Whether the file has the correct type

File validation helps prevent runtime errors.

```python
import os

filename = "report.txt"

if os.path.exists(filename):
    print("File is available")
else:
    print("File is not available")
```

```python
import os

filename = "report.txt"

if os.path.isfile(filename):
    print("This is a file")
else:
    print("This is not a file")
```

```python
import os

filename = "report.txt"

if os.path.exists(filename):
    with open(filename, "r") as file:
        content = file.read()

    if content.strip():
        print("File contains data")
        print(content)
    else:
        print("File is empty")
else:
    print("File does not exist")
```

## 11) DIRECTORY OPERATIONS

Directory operations are used to manage folders.

Common operations:

Creating folders

Deleting folders

Renaming folders

Listing folders

```python
import os

folder = "MyFiles"

if not os.path.exists(folder):
    os.mkdir(folder)
    print("Directory created")
else:
    print("Directory already exists")
```

```python
import os

old_name = "MyFiles"
new_name = "MyDocuments"

if os.path.exists(old_name):
    os.rename(old_name, new_name)
    print("Directory renamed")
else:
    print("Directory not found")
```
