# Module 4 - LOGGING

## 1. What is Logging?

Logging means recording information about what happens when a Python program runs.

It can record:

- normal activities
- warnings
- errors
- important events

Logging helps us understand what happened in a program.

```python
import logging

logging.warning("Something needs attention")
```

## 2. Why is Logging used?

Logging is used to:

- find errors
- check program flow
- monitor the program
- save important information
- understand what happened during execution

Logging is more useful than print() when working with bigger programs.

## 3. Logging and print()

print() is mainly used to display information on the screen.

logging is used to record information about the program.

print() is useful for simple output.

logging is useful for debugging and monitoring.

```python
print("Hello Python")

import logging

logging.warning("This is a warning")
```

## 4. Creating folders for logs

We can create a separate folder to store log files.

The logs folder will contain the log files created by the program.

```python
import os

os.makedirs("logs", exist_ok=True)

print("Log folder created")
```

## 5. Importing the logging module

Python provides a built-in module called logging.

We can import it using the import statement.

```python
import logging

print("Logging module imported successfully")
```

## 6. Creating a Log File

We can save log messages inside a file.

filename is used to specify the log file.

level decides which log messages should be recorded.

format decides how the message will be displayed.

```python
import logging

logging.basicConfig(
    filename="logs/mylog.log",
    level=logging.INFO
)

logging.info("My first log message")

print("Log message added")
```

## 7. Log Levels

Logging has different levels based on the importance of a message.

DEBUG - detailed information

INFO - normal information

WARNING - something unexpected happened

ERROR - an error occurred

CRITICAL - a serious problem occurred

Order:

DEBUG < INFO < WARNING < ERROR < CRITICAL

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug message")
logging.info("Information message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

## 8. Using Different Log Levels

We can use different logging methods depending on the situation.

debug() is used for debugging.

info() is used for normal information.

warning() is used for warnings.

error() is used for errors.

critical() is used for serious errors.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Checking the program")
logging.info("Program started")
logging.warning("Check the value")
logging.error("Something went wrong")
logging.critical("Serious problem")
```

## 9. Formatter

A formatter decides how the log message looks.

A log message can contain:

- date and time
- log level
- message

A formatter makes logs easier to read.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s - %(message)s"
)

logging.info("Login successful")
logging.warning("Wrong password")
```

## 10. Creating a Formatter

We can use different formatting options.

%(asctime)s - date and time

%(levelname)s - log level

%(message)s - log message

```python
import logging

logger = logging.getLogger("my_logger")

logger.setLevel(logging.INFO)

handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

handler.setFormatter(formatter)

logger.addHandler(handler)

logger.info("Welcome to Python")
```

# 11. Handlers

A handler decides where the log message goes.

A log message can be sent to:

- screen
- file

```python
import logging

logger = logging.getLogger("example")

logger.setLevel(logging.INFO)

handler = logging.StreamHandler()

logger.addHandler(handler)

logger.info("Hello from logger")
```

# 12. File Handler

A FileHandler saves log messages into a file.

```python
import logging

logger = logging.getLogger("file_example")

logger.setLevel(logging.INFO)

handler = logging.FileHandler("logs/file.log")

logger.addHandler(handler)

logger.info("This message is saved in the file")

print("Log saved")
```

# 13. Console and File Handler

We can use a console handler and a file handler together.

Console handler shows the message on the screen.

File handler saves the message in a file.

```python
import logging

logger = logging.getLogger("both")

logger.setLevel(logging.INFO)

console = logging.StreamHandler()
file = logging.FileHandler("logs/both.log")

logger.addHandler(console)
logger.addHandler(file)

logger.info("Hello")
logger.warning("Be careful")
```

# 14. Rotating Logs

A log file can become very large.

Rotating logs help create backup files when the log file becomes too large.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotate")

logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "logs/rotate.log",
    maxBytes=100,
    backupCount=2
)

logger.addHandler(handler)

logger.info("This is a rotating log example")

print("Rotating log created")
```

# 15. maxBytes and backupCount

maxBytes tells the maximum size of the log file.

backupCount tells how many backup files should be kept.

```python
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "logs/test.log",
    maxBytes=100,
    backupCount=2
)

print("Maximum size:", handler.maxBytes)
print("Backup count:", handler.backupCount)
```

# 16. Checking Log File

We can check whether a log file exists using os.path.exists().

```python
import os

if os.path.exists("logs/mylog.log"):
    print("Log file exists")
else:
    print("Log file not found")
```

# 17. Reading a Log File

We can open a log file and read the messages stored inside it.

```python
import os
import logging

os.makedirs("logs", exist_ok=True)

logging.basicConfig(
    filename="logs/mylog.log",
    level=logging.INFO,
    force=True
)

logging.info("Program started")
logging.info("This is my first log message")

print("Log file created")
```

```python
with open("logs/mylog.log", "r") as file:
    data = file.read()

print(data)
```
