# MODULE 3 - EXCEPTION HANDLING

## What is Exception Handling?

Exception Handling is used to handle errors that occur while a Python program is running.

It prevents the program from stopping suddenly when an error occurs.

Python provides different keywords for handling exceptions:

```text
try       - Contains the code that may cause an error
except    - Handles the error
else      - Runs when there is no error
finally   - Always runs whether an error occurs or not
raise     - Used to create an exception manually
assert    - Used to check a condition
```

Exception handling helps us write programs that can handle errors properly and continue running safely.

try is used to write the code that may cause an error.

If an error occurs inside the try block, Python will move to the except block.

## 1) try BLOCK

The `try` block is used to write the code that may cause an error.

If an error occurs inside the `try` block, Python will move to the `except` block.

### Syntax:

```python
try:
    # code that may cause an error
```


````python
try:
    number = int(input("Enter a number: "))
    print("Number:", number)

except:
    print("Invalid input")
````

## 2) except BLOCK

The `except` block is used to handle an error.

When an error occurs in the `try` block, the `except` block will execute.

It prevents the program from stopping suddenly.


````python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Please enter a valid number")
````

## 3) try with except

We can use `try` and `except` together to handle errors.

The `try` block contains the code that may produce an error.

The `except` block handles the error.


````python
try:
    a = 10
    b = 0
    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
````

## 4) else BLOCK

The `else` block is executed when there is no error in the `try` block.

The `else` block runs only when the `try` block is successful.


````python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid number")

else:
    print("You entered:", number)
````

## 5) finally BLOCK

The `finally` block always executes.

It runs whether an error occurs or not.

It is commonly used for cleanup operations.


````python
try:
    number = 10
    result = number / 2
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Program completed")
````

## 6) raise KEYWORD

The `raise` keyword is used to create an exception manually.

We can use `raise` when we want to stop the program and show an error message for an invalid condition.


````python
age = int(input("Enter your age: "))

if age < 18:
    raise ValueError("Age must be 18 or above")

print("You are eligible")
````

## 7) assert KEYWORD

The `assert` keyword is used to check whether a condition is true.

If the condition is true, the program continues.

If the condition is false, Python raises an `AssertionError`.


````python
age = 20

assert age >= 18

print("Age is valid")
````

````python
age = 15

assert age >= 18, "Age must be 18 or above"

print("Age is valid")
````

## 7) CUSTOM EXCEPTIONS

A custom exception is an exception created by the programmer.

We create a custom exception by using a class and Exception.

````python
class AgeError(Exception):
    pass


try:
    age = 15

    if age < 18:
        raise AgeError("Age must be 18 or above")

except AgeError as e:
    print("Error:", e)
````
